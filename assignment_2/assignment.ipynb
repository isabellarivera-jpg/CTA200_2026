# Part 1

### Write a python function for the function f(x) = x^3 - x^2 - 1.

```python
def f(x):
    """
    This function evaluates the polynomial f(x) = x**3 - x**2 - 1.

    Parameters
    ----------
    x : float or array-like
        The input value(s) at which to evaluate the polynomial.

    Returns
    -------
    y : float or array-like
        The value of x**3 - x**2 - 1 evaluated at x.
    """
    y = x**3 - x**2 - 1
    
    return y
```

```python
assert f(0) == -1, f"Expected -1, got {f(0)}"
assert f(2) == 3,  f"Expected 3, got {f(2)}"
assert f(-2) == -13, f"Expected -13, got {f(-2)}"
```

### Write a function for the derivative of the function f(x) = x^3 - x^2 - 1.

```python
def df(x):
    """
    Evaluates the derivative of f(x) = x**3 - x**2 - 1.

    Parameters
    ----------
    x : float
        The input value at which to evaluate the derivative.

    Returns
    -------
    dy : float
        The value of f'(x) = 3*x**2 - 2*x evaluated at x.
    """
    dy = 3*(x**2) - 2*x

    return dy
```

```python
assert df(0) == 0,   f"Expected 0, got {df(0)}"
assert df(2) == 8,   f"Expected 8, got {df(2)}"
assert df(-2) == 16, f"Expected 16, got {df(-2)}"
```

# Part 2

### Create a function that performs a Newton Iteration to find the root of f using its derivative df.

```python
def newton(f, df, x0, epsilon=1e-6, max_iter=30):
    """
    Finds a root of f using Newton's method.
    
    Parameters
    ----------
    f : callable
        The function whose root we want to find.
    df : callable
        The derivative of f.
    x0 : float
        Initial guess for the root.
    epsilon : float, optional
        Convergence tolerance. Stops when |f(xn)| < epsilon. Default is 1e-6.
    max_iter : int, optional
        Maximum number of iterations allowed. Default is 30.

    Returns
    -------
    xn : float or None
        The approximate root if convergence is achieved, otherwise None.
    """
    xn = x0 #setting the initial place 
    for i in range(max_iter): #looping through the iterations
        fxn = f(xn) 
        if abs(fxn) < epsilon: #condition when to exit loop
            print(f"Found root in {i} iterations")
            return xn
        xn = xn - fxn / df(xn) #recursive 

    print("Iteration failed")
    return None
```

# Part 3

### Testing out the Newton Iteration function

```python
# Example 1: starting the guess close to the root
root1 = newton(f, df, x0=2.0)
print(f"Root: {root1}, f(root) = {f(root1):.2e}")

# Example 2: starting the guess further away
root2 = newton(f, df, x0=7)
print(f"Root: {root2}, f(root) = {f(root2):.2e}")
```

```python
# Reducing epsilon to 1e-8
root3 = newton(f, df, x0=2.0, epsilon=1e-8)
print(f"Root: {root3}, f(root) = {f(root3):.2e}")
```

With `epsilon=1e-6`, the method converges in 4 iterations starting from x0=2.0. 
Reducing `epsilon` to `1e-8` still works, converging in 5 iterations, so only 1 
extra iteration. This is expected because Newton's method has quadratic 
convergence, meaning each iteration roughly doubles the number of correct 
decimal places, so only a small number of extra iterations are needed to gain 
the additional precision.
