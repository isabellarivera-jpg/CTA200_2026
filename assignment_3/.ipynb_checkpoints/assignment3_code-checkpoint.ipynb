```python
import numpy as np
import matplotlib.pyplot as plt
```

## Question 1

```python
#importing the function I made that is now in the .py file
from function_q1 import iterate
```

```python
# define the range
x = np.linspace(-2, 2.0, 1000)
y = np.linspace(-2.0, 2.0, 1000)
X, Y = np.meshgrid(x, y)

# applying the function I made
diverged, iter_count = iterate(X, Y, max_iter=100)

fig, axes = plt.subplots(1, 2, figsize=(25, 8))

# font sizes
title_fs = 30
label_fs = 28
tick_fs = 24
cbar_fs = 24
suptitle_fs = 34

# Plot 1: Bounded vs Diverged
im1 = axes[0].imshow(
    diverged,
    extent=[-2, 2, -2, 2],
    cmap='PuRd',
    origin='lower'
)

cbar1 = fig.colorbar(im1, ax=axes[0], ticks=[0, 1], fraction=0.046, pad=0.04)
cbar1.ax.set_yticklabels(['Bounded', 'Diverged'], fontsize=cbar_fs)
cbar1.ax.tick_params(labelsize=cbar_fs)

axes[0].set_title('Bounded vs Diverged', fontsize=title_fs)
axes[0].set_xlabel('Real', fontsize=label_fs)
axes[0].set_ylabel('Imaginary', fontsize=label_fs)
axes[0].tick_params(axis='both', labelsize=tick_fs)

axes[0].set_xlim(-2, 0.5)
axes[0].set_ylim(-1.25, 1.25)

# Plot 2: Iteration Count
im2 = axes[1].imshow(
    iter_count,
    extent=[-2, 2, -2, 2],
    cmap='PuRd',
    origin='lower'
)

axes[1].set_xlim(-2, 0.5)
axes[1].set_ylim(-1.25, 1.25)

cbar2 = fig.colorbar(im2, ax=axes[1], fraction=0.046, pad=0.04)
cbar2.ax.tick_params(labelsize=cbar_fs)
cbar2.set_label('Iteration count', fontsize=cbar_fs)

axes[1].set_title('Iteration Count', fontsize=title_fs)
axes[1].set_xlabel('Real', fontsize=label_fs)
axes[1].set_ylabel('Imaginary', fontsize=label_fs)
axes[1].tick_params(axis='both', labelsize=tick_fs)

plt.tight_layout()

plt.savefig('Mandelbrot_Set.pdf', bbox_inches='tight')
plt.show()
```

### Part 2.2

```python
dt = 0.01
t_eval = np.arange(0, 30, dt)  # 3000 iterations total
t_range = [0, 30]
p, r, b = 10., 28, 8./3.

sol = solve_ivp(lorenz, t_range, initial_cond, args = (p, r, b), t_eval = t_eval)

Y = sol.y[1]
N = sol.t / dt  # convert time to iteration number

fig, axes = plt.subplots(3, 1, figsize=(12, 8))

#splitting up like how Lorenz did
chunks = [(0, 1000), (1000, 2000), (2000, 3000)]

#plotting
for i, (start, end) in enumerate(chunks):
    
    axes[i].plot(N[start:end], Y[start:end], color='hotpink')
    axes[i].set_ylabel('Y', fontsize=18)
    axes[i].tick_params(labelsize=15)
    axes[i].axhline(0, color='black', lw=0.5)
    axes[i].set_xlim(N[start], N[end-1])
    axes[i].grid(color='gray', linestyle='--', linewidth=0.5)


axes[2].set_xlabel('N', fontsize=18)
fig.suptitle('Lorenz Figure 1 Reproduction', fontsize=22)
plt.tight_layout()
plt.savefig('Lorenz_1.pdf', bbox_inches='tight')
plt.show()
```

### Part 2.3

```python
sol2 = solve_ivp(
    lorenz,
    [0, 60],
    initial_cond,
    args=(p, r, b),
    max_step=0.01,
    dense_output=True
)

t_fine = np.linspace(14, 19, 1000)
W = sol2.sol(t_fine)
X2 = W[0]
Y2 = W[1]
Z2 = W[2]

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

axes[0].plot(Y2, Z2, color='hotpink')
axes[0].set_xlabel('Y', fontsize=18)
axes[0].set_ylabel('Z', fontsize=18)
axes[0].set_title('Y vs Z', fontsize=20)
axes[0].tick_params(labelsize=15)
axes[0].grid(True)

axes[1].plot(Y2, X2, color='hotpink')
axes[1].set_xlabel('Y', fontsize=18)
axes[1].set_ylabel('X', fontsize=18)
axes[1].set_title('Y vs X', fontsize=20)
axes[1].tick_params(labelsize=15)
axes[1].invert_yaxis()
axes[1].grid(True)

fig.suptitle('Lorenz Figure 2 Reproduction', fontsize=22)
plt.tight_layout()
plt.savefig('Lorenz_2.pdf', bbox_inches='tight')
plt.show()
```

```python
# solve with slightly different initial conditions
initial_cond2 = [0., 1.00000001, 0.]
sol3 = solve_ivp(
    lorenz,
    [0, 60],
    initial_cond2,
    args=(p, r, b),
    max_step=0.01,
    dense_output=True
)
# evaluate both solutions at the same time points
t_common = np.arange(0, 60, 0.01)
W1 = sol2.sol(t_common)
W2 = sol3.sol(t_common)
# compute distance between the two solutions
distance = np.sqrt((W1[0]-W2[0])**2 + (W1[1]-W2[1])**2 + (W1[2]-W2[2])**2)
# semilog plot
plt.figure(figsize=(10, 6))
plt.ylim(1e-8, 1e2)
plt.semilogy(t_common, distance, color='hotpink')
plt.xlabel(r'Time $t$', fontsize=18)
plt.ylabel(r"Distance $\|W' - W\|$ (log scale)", fontsize=18)
plt.tick_params(labelsize=15)
plt.title('Sensitivity to Initial Conditions', fontsize=20)
plt.grid(True)
plt.tight_layout()
plt.savefig('Sensitivity.pdf', bbox_inches='tight')
plt.show()
```
